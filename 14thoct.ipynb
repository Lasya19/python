```python
n=int(input())
for i in range(n):
    for j in range(n):
        if i==0 or j==0 or i==n-1 or j==n-1:
            print("*",end=" ")
        else:
            print(" ",end=" ")
    print()
```

```python
b=input()
vowels=['a','e','i','o','u']
v=[]
c=[]
for l in b:
  if l in vowels:
    v.append(l)
  else:
    c.append(l)
print(f"vowels are {v} - {len(v)}")
print(f"consants are {c} - {len(c)}")
```

```python
year=list(map(int,input().split()))
leapyear=[]
nonleapyear=[]
for i in year:
  if(i%4==0 and i%100!=0) or (i%400==0):
    leapyear.append(i)
  else:
    nonleapyear.append(i)
print(leapyear)
print(nonleapyear)
```
