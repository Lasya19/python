```python
# ★Write a Python program to reverse a
# given string.
p = input()
i = len(p) - 1
rev = ""
while i >= 0:
    rev += p[i]
    i -= 1

print(rev)
```

```python
# ★Count Vowels in a String
l=["a","e","i","o","u"]
q=input()
count=0
for i in range(len(q)):
  if q[i] in l:
    count+=1

print(count)
```

```python
# ★Count consonants in a string
l=["a","e","i","o","u"]
r=input()
count=0
for i in range(len(r)):
  if r[i] not in l:
    count+=1

print(count)
```
