```python
#translation
from gtts import gTTS
import speech_recognition as sr
import pyaudio
import os
from googletrans import Translator
#speech - text
def ListentText(): 
    record=sr.Recognizer()
    with sr.Microphone() as voice:
        print("start speaking...")
        audio=record.listen(voice)
    info=""
    info=record.recognize_google(audio)
    print(info)
    #translation:
    t=Translator()
    converted=t.translate(text=info,dest="hi")
    print(converted.text)
    #text - speech
    tts=gTTS(text=converted.text)
    try:
     tts.save("base.mp3")
     os.startfile("base.mp3")
     print("opened diretly")
     n=int(input("to close enter any number:"))
    finally:
     os.remove("base.mp3")
     print("file deleted succefully")
    print("success")



ListentText()
```
