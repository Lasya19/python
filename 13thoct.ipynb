```python
n=int(input())
i=0
while i<n:
  spaces=0
  while spaces<n-i-1:
    print(" ",end=" ")
    spaces+=1
  stars=0
  while stars<i+1:
    print("*",end=" ")
    stars+=1
  print()
  i+=1
```

```python
n=4
rows=0
while rows<n:
    spaces=0
    while spaces<n-rows-1:
        print(" ",end=" ")
        spaces+=1

    stars=0
    while stars<rows+1:
        print(stars,end=' ')
        stars+=1
    print()
    rows+=1
```
